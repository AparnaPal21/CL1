# Practical 1: Feature Transformation - PCA on the Wine dataset

Standardize the 13 wine measurements, reduce them to 2 principal components with PCA, and plot the three wine classes in that 2-D space.

```python
# Where the dataset lives:
#  - Google Colab: mounts Google Drive and reads from MyDrive/Data-Science
#  - Local Jupyter: put the CSV in the same folder as this notebook
import os
try:
    from google.colab import drive
    drive.mount('/content/drive')
    DATA_DIR = '/content/drive/MyDrive/Data-Science'
except ImportError:
    DATA_DIR = '.'
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
WINE_CSV = os.path.join(DATA_DIR, 'Wine dataset.csv')

if os.path.exists(WINE_CSV):
    df = pd.read_csv(WINE_CSV)
else:
    # Same UCI wine data, built from scikit-learn with the same column names as "Wine dataset.csv"
    print("'Wine dataset.csv' not found - using the copy bundled with scikit-learn.")
    from sklearn.datasets import load_wine
    wine = load_wine()
    df = pd.DataFrame(wine.data, columns=['Alcohol', 'Malic acid', 'Ash', 'Alcalinity of ash', 'Magnesium',
                                          'Total phenols', 'Flavanoids', 'Nonflavanoid phenols',
                                          'Proanthocyanins', 'Color intensity', 'Hue',
                                          'OD280/OD315 of diluted wines', 'Proline'])
    df.insert(0, 'class', wine.target + 1)
    df['Magnesium'] = df['Magnesium'].astype(int)
    df['Proline'] = df['Proline'].astype(int)
df.head()
```

```python
# Separate features (X) and target (y)
X = df.drop('class', axis=1)
y = df['class']

# Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Shape of scaled features:", X_scaled.shape)
print("First 5 rows of scaled features (sample):\n", X_scaled[:5])
```

```python
# Apply PCA to reduce to 2 components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Create a DataFrame for the PCA results
pca_df = pd.DataFrame(data = X_pca, columns = ['principal component 1', 'principal component 2'])

# Add the target variable back for visualization
pca_df['class'] = y

print("Shape of PCA transformed data:", X_pca.shape)
display(pca_df.head())
```

```python
plt.figure(figsize=(10, 8))
sns.scatterplot(
    x='principal component 1',
    y='principal component 2',
    hue='class',
    data=pca_df,
    palette='viridis',
    s=100,  # Marker size
    alpha=0.8  # Transparency
)

plt.title('2 Component PCA of Wine Dataset', fontsize=16)
plt.xlabel('Principal Component 1', fontsize=12)
plt.ylabel('Principal Component 2', fontsize=12)
plt.grid(True)
plt.legend(title='Wine Class')
plt.show()
```
